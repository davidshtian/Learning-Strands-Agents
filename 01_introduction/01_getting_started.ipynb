## Your First Agent

Creating a Strands agent is straightforward. Here's how to create a simple agent that can respond to queries:

```python
from strands import Agent

# Create a simple agent
simple_agent = Agent(
    model="us.amazon.nova-lite-v1:0",  # Using Nova Lite model
    system_prompt="You are a helpful assistant that provides clear and informative responses."
)

# Use the agent by calling it with a query
response = simple_agent("What is machine learning and why is it important?")
```

## Understanding Agent Responses

When you call an agent with a query, it returns an `AgentResult` object. This object contains:

- The message from the agent (`response.message`)
- The metrics from the agent (`response.metrics`)

Let's examine the response in more detail:

```python
# Let's try another query
response = simple_agent("who are you?")

print("\n" + "-"*80 + "\n")
print("Response Type:")
print(type(response))

# Print the message
print("\n" + "-"*80 + "\n")
print("Response Message:")
print(response.message)

# Print the metrics
print("\n" + "-"*80 + "\n")
print("Response Metrics:")
print(response.metrics)
```

## Customizing Your Agent

One of the powerful features of Strands Agents is the ability to customize their behavior through system prompts. The system prompt defines the agent's role, personality, and constraints.

Let's create a few agents with different personalities:

```python
# A technical expert agent
tech_expert = Agent(
    model="us.amazon.nova-lite-v1:0",
    system_prompt="""You are a senior software engineer with deep expertise in computer science.
    You provide detailed, technically accurate explanations using precise terminology.
    When appropriate, you include code examples to illustrate concepts.
    You focus on best practices and modern approaches to software development."""
)

# An educational agent for beginners
beginner_tutor = Agent(
    model="us.amazon.nova-lite-v1:0",
    system_prompt="""You are a patient and encouraging tutor for beginners.
    You explain complex topics using simple language and relatable analogies.
    You break down information into manageable chunks and avoid technical jargon.
    You focus on building confidence and understanding foundational concepts."""
)

# A creative storyteller agent
storyteller = Agent(
    model="us.amazon.nova-lite-v1:0",
    system_prompt="""You are a creative storyteller with a vivid imagination.
    You transform ordinary topics into engaging narratives with colorful characters.
    You use descriptive language, metaphors, and narrative structures to make information memorable.
    You maintain a sense of wonder and curiosity in your responses."""
)
```

Now let's ask each agent the same question and compare their responses:

```python
question = "How does the internet work?"

# Get responses from each agent

# Print the responses
print("Technical Expert's Response:\n")
tech_response = tech_expert(question)

print("\n" + "-"*80 + "\n")

print("Beginner Tutor's Response:\n")
beginner_response = beginner_tutor(question)

print("\n" + "-"*80 + "\n")

print("Storyteller's Response:\n")
story_response = storyteller(question)
```

## Multi-turn Conversations

Strands agents can maintain context across multiple turns of conversation. Let's see how this works:

```python
# Create an agent for conversation
conversation_agent = Agent(
    model="us.amazon.nova-lite-v1:0",
    system_prompt="""You are a helpful assistant that remembers previous parts of the conversation.
    You respond appropriately to follow-up questions and references to earlier topics.
    You are concise but informative."""
)

# Start a conversation
print("\nFirst turn:")
response1 = conversation_agent("Tell me about three famous machine learning algorithms.")

print("\n" + "-"*80 + "\n")
print("Second turn (follow-up):")
response2 = conversation_agent("Which of these is best for image classification?")

print("\n" + "-"*80 + "\n")
print("Third turn (follow-up):")
response3 = conversation_agent("Can you provide an example use case for the first algorithm you mentioned?")
```

## Agent Configuration Options

When creating agents, you can configure several parameters to customize their behavior. Let's explore some of these options:

```python
from strands.models import BedrockModel

# Create a BedrockModel with your desired parameters
bedrock_model = BedrockModel(
    model_id="us.amazon.nova-lite-v1:0",  # Your model ID
    region_name='us-east-1',  # Specify your AWS region
    temperature=0.7,          # Model-specific parameters go here
    max_tokens=500,
    top_p=0.9
)

# Create agent with the configured model
custom_agent = Agent(
    model=bedrock_model,
    system_prompt="You are a helpful assistant."
)

# Test the agent
response = custom_agent("Write a short poem about artificial intelligence.")
```

## Handling Input Context

Sometimes you want to provide additional context to the agent beyond the immediate query. This can be useful for tasks like summarization, analysis, or working with external content:

```python
# Create an agent for analyzing text
analyzer_agent = Agent(
    model="us.amazon.nova-lite-v1:0",
    system_prompt="""You are a text analysis specialist.
    You can analyze provided text for key points, sentiment, themes, and structure.
    You provide clear, organized analysis with specific examples from the text."""
)

# Sample text to analyze
sample_article = """
The Evolution of Artificial Intelligence

Artificial Intelligence has undergone a remarkable transformation since its conceptual birth in the 1950s. What began as rule-based systems and simple algorithms has evolved into sophisticated neural networks capable of outperforming humans in specific tasks.

The early days of AI were characterized by optimism and grand promises. Researchers believed that human-level intelligence was just around the corner. However, this period was followed by the infamous "AI winter," where funding dried up as early promises failed to materialize.

The resurgence of AI in the 21st century can be attributed to three key factors: the explosion of available data, significant improvements in computing power, and algorithmic innovations. Deep learning, a subset of machine learning based on artificial neural networks, has been particularly transformative.

Today's AI systems can recognize images, understand natural language, play complex games, and even create art. However, they still lack the general intelligence and common sense reasoning that humans possess naturally. The field continues to grapple with challenges related to ethics, bias, and explainability.

As we look to the future, the development of artificial general intelligence remains a tantalizing goal. Whether this will lead to utopian advancements or existential risks is a subject of intense debate among researchers, philosophers, and policymakers alike.
"""

# Ask the agent to analyze the text
response = analyzer_agent(f"Please analyze this article on AI evolution:\n\n{sample_article}")
```

## Introduction to Tools

One of the most powerful features of Strands Agents is their ability to use tools. Tools are functions that agents can call to perform actions, retrieve information, or interact with external systems.

In the next chapter, we'll explore tools in depth, but let's get a quick introduction here:

```python
from strands import Agent, tool
from strands_tools import calculator

# Define a simple custom tool
@tool
def get_current_time() -> str:
    """
    Get the current time and date.
    
    Returns:
        str: The current time and date.
    """
    import datetime
    now = datetime.datetime.now()
    return f"The current time is {now.strftime('%H:%M:%S')} on {now.strftime('%Y-%m-%d')}"

# Create an agent that can use tools
tool_using_agent = Agent(
    model="us.amazon.nova-lite-v1:0",
    tools=[calculator, get_current_time],  # Give the agent access to tools
    system_prompt="""You are a helpful assistant that can perform calculations and check the current time.
    Use the calculator tool for mathematical operations.
    Use the get_current_time tool to find out the current time and date."""
)

# Ask the agent to use its tools
response = tool_using_agent("What is 1234 * 5678 and what time is it now?")
```

## Best Practices for Effective Agents

To create effective Strands agents, keep these best practices in mind:

1. **Clear System Prompts**: Provide detailed and specific instructions in your system prompt
2. **Focused Agents**: Create agents with well-defined responsibilities rather than trying to do everything
3. **Appropriate Tools**: Equip agents with tools that align with their intended purpose
4. **Error Handling**: Anticipate and handle potential errors in your agent implementation
5. **Testing**: Thoroughly test your agents with various inputs to ensure they behave as expected
6. **Iterative Improvement**: Refine your agents based on their performance and user feedback
