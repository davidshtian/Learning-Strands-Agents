## Categories of Built-in Tools

The `strands_tools` package categorizes tools into several groups based on functionality:

1. **Utility Tools**: Basic operations like calculations, time, etc.
2. **File Operations**: Read, write, and manipulate files
3. **Code Interpretation**: Run code and evaluate expressions
4. **Shell & System**: Execute shell commands and manage environment
5. **Web & Network**: Make HTTP requests and work with web data
6. **Multi-modal**: Process images and generate media
7. **AWS Services**: Interact with AWS services
8. **RAG & Memory**: Retrieve and manage knowledge
9. **Agents & Workflows**: Coordinate complex agent systems

Let's explore each category with examples.

## 1. Utility Tools

### Calculator Tool

The `calculator` tool allows agents to perform mathematical calculations. Let's start by using this simple but powerful tool:

```python
from strands import Agent
from strands_tools import calculator

# Create an agent with the calculator tool
calculator_agent = Agent(
    model="us.amazon.nova-lite-v1:0",
    tools=[calculator],
    system_prompt="You are an assistant that can perform mathematical calculations."
)

# Ask the agent to perform calculations
response = calculator_agent("""
Can you help me with these calculations:
1. Calculate 1234 * 5678
2. What is the square root of 169?
3. Calculate (15^3 + 27) / 42
""")
```

### Current Time Tool

The `current_time` tool provides the current date and time information:

```python
from strands_tools import current_time

# Create an agent with the current_time tool
time_agent = Agent(
    model="us.amazon.nova-lite-v1:0",
    tools=[current_time],
    system_prompt="You are an assistant that provides accurate time information."
)

# Ask for time information
response = time_agent("""
I need the following information:
1. What is the current date and time?
2. What day of the week is today?
""")
```

## 2. File Operations

Strands provides tools for reading from and writing to files, which are essential for many tasks.

### File Read Tool

Let's first create a file to read, then use the `file_read` tool to access its contents:

```python
# Create a sample file
with open("sample_data.txt", "w") as f:
    f.write("""# Sample Data File
This is a sample data file with some information.

## Product Sales
Product A: 145 units sold at $29.99 each
Product B: 89 units sold at $49.99 each
Product C: 213 units sold at $19.99 each

## Website Traffic
January: 12,450 visitors
February: 14,280 visitors
March: 18,765 visitors
""")

print("Sample file created successfully!")
```

```python
from strands_tools import file_read

# Create an agent with the file_read tool
file_reader_agent = Agent(
    model="us.amazon.nova-lite-v1:0",
    tools=[file_read, calculator],
    system_prompt="You are an assistant that can read files and analyze their content."
)

# Ask the agent to read and analyze the file
response = file_reader_agent("""
Please read the file 'sample_data.txt', answer these questions:
1. What products are mentioned in the file?
2. What was the total revenue from all products?
3. What was the average website traffic over the three months?
""")
```

### File Write Tool

The `file_write` tool allows agents to create or modify files:

```python
from strands_tools import file_write

# Create an agent with the file_write tool
file_writer_agent = Agent(
    model="us.amazon.nova-lite-v1:0",
    tools=[file_write, calculator],
    system_prompt="You are an assistant that can create files with useful content."
)

# Ask the agent to create a file with a sales report
response = file_writer_agent("""
Based on this data:
- Product X: 120 units at $25 each
- Product Y: 85 units at $42 each
- Product Z: 200 units at $15 each

Create a file named 'sales_report.md', with a professional sales report that includes:
1. A title and introduction
2. A breakdown of each product's sales
3. Total revenue
4. Average price per unit
5. Best-performing product
""")
```

Let's verify the file was created and check its contents:

```python
# Read the generated sales report
with open("sales_report.md", "r") as f:
    content = f.read()
    
print(content)
```

## 3. Code Interpretation

### Python REPL Tool

The `python_repl` tool allows agents to execute arbitrary Python code, which is incredibly powerful for data processing, calculations, and more:

```python
from strands_tools import python_repl

os.environ["PYTHON_REPL_INTERACTIVE"] = "False"

# Create an agent with the python_repl tool
python_agent = Agent(
    model="us.amazon.nova-lite-v1:0",
    tools=[python_repl],
    system_prompt="You are an assistant that can execute efficient Python code to solve problems."
)

# Ask the agent to solve problems using Python
response = python_agent("""
books = [
    {"title": "The Hobbit", "genre": "Fantasy"},
    {"title": "Gone Girl", "genre": "Mystery"},
    {"title": "Dune", "genre": "Sci-Fi"},
    {"title": "The Da Vinci Code", "genre": "Mystery"},
    {"title": "The Lord of the Rings", "genre": "Fantasy"}
]

Organize books by genre for me.
""")
```

## 5. Web & Network Operations

### HTTP Request Tool

The `http_request` tool allows agents to make HTTP requests, which is useful for interacting with web services and APIs:

```python
from strands_tools import http_request

os.environ["PYTHON_REPL_INTERACTIVE"] = "False"

# Create an agent with the http_request tool
http_agent = Agent(
    model="us.amazon.nova-lite-v1:0",
    tools=[http_request, python_repl],
    system_prompt="You are an assistant that can make HTTP requests to gather information."
)

# Ask the agent to make HTTP requests
response = http_agent("""
1. Make a GET request to https://jsonplaceholder.typicode.com/todos/1 and show me the response
2. Make a GET request to fetch a list of users from https://jsonplaceholder.typicode.com/users and extract their names and email addresses
3. Use Python to create a nicely formatted table of the users' information
""")
```

## 6. Multi-modal Tools

### Image Reader Tool

The `image_reader` tool allows agents to analyze images. Let's create a simple image first:

```python
# Create a simple image for demonstration
from PIL import Image, ImageDraw, ImageFont
import numpy as np

# Create a blank image
img = Image.new('RGB', (500, 300), color='white')
d = ImageDraw.Draw(img)

# Add some text
try:
    font = ImageFont.truetype("Arial.ttf", 36)
except IOError:
    font = ImageFont.load_default()
    
d.text((50, 50), "Strands Agents", fill=(0, 0, 0), font=font)
d.text((50, 120), "Multi-modal Processing", fill=(0, 0, 200), font=font)

# Add some shapes
d.rectangle([(50, 200), (200, 250)], outline="red", width=3)
d.ellipse([(300, 180), (450, 270)], outline="green", width=3)

# Save the image
img.save("test_image.png")

# Display the image in the notebook
display(img)
```

```python
from strands_tools import image_reader

# Create an agent with the image_reader tool
image_agent = Agent(
    model="us.amazon.nova-lite-v1:0",
    tools=[image_reader],
    system_prompt="You are an assistant that can analyze images."
)

# Ask the agent to analyze the image
response = image_agent("""
Analyze the image file 'test_image.png' and tell me:
1. What text is present in the image?
2. What shapes are drawn in the image?
3. What colors are used?
""")
```

## Combining Multiple Tools

In real-world applications, agents typically need to use multiple tools together to solve complex problems. Let's create an agent with several tools and ask it to solve a more complex task:

```python
from strands import Agent
from strands_tools import calculator, current_time, python_repl, file_read, file_write, http_request

# Create a multi-tool agent
multi_tool_agent = Agent(
    model="us.amazon.nova-lite-v1:0",
    tools=[
        current_time,
        python_repl,
        file_read,
        file_write,
        http_request
    ],
    system_prompt="You are a versatile assistant that can solve complex problems using multiple tools."
)

# Ask the agent to solve a complex task using multiple tools
response = multi_tool_agent("""
I need you to complete a data analysis task:

1. Fetch user data from https://jsonplaceholder.typicode.com/users
2. Use Python code to find the user who lives furthest north and furthest south
3. Create a file called 'user_analysis.md' with a summary of your findings
4. Include the current date and time in the report
""")
```

## Tool Usage Patterns

Now that we've explored several built-in tools, let's discuss some best practices and patterns for using tools effectively:

### 1. Use Clear System Prompts

When creating an agent with tools, a clear system prompt helps guide how the tools are used:

```python
# Example of a good system prompt for tools
good_system_prompt = """
You are an assistant that helps with data analysis.
When presented with data, follow these steps:
1. Use tools to read and understand the data format
2. Clean and process the data using Python when needed
3. Perform calculations or analysis as requested
4. Present results clearly with explanations

Always explain your reasoning before using tools. When using Python, write clean, well-commented code.
"""

# Create an agent with this system prompt
data_agent = Agent(
    model="us.amazon.nova-lite-v1:0",
    tools=[file_read, python_repl, calculator],
    system_prompt=good_system_prompt
)
```

### 2. Tool Selection Strategy

Only include tools that are relevant to the agent's purpose. Too many unnecessary tools can distract the agent and lead to suboptimal solutions.

```python
# Example: Creating specialized agents with focused tool sets

# Data processing agent
data_agent = Agent(
    model="us.amazon.nova-lite-v1:0",
    tools=[file_read, file_write, python_repl],
    system_prompt="You are specialized in data processing and analysis."
)

# System operations agent
system_agent = Agent(
    model="us.amazon.nova-lite-v1:0",
    tools=[shell, environment],
    system_prompt="You are specialized in system operations and environment management."
)

# Web interaction agent
web_agent = Agent(
    model="us.amazon.nova-lite-v1:0",
    tools=[http_request, python_repl],
    system_prompt="You are specialized in web data retrieval and processing."
)
```

### 3. Tool Error Handling

It's important to understand that tools can fail or produce unexpected results. Let's look at an example of error handling:

```python
# Create an agent that might encounter errors
error_handling_agent = Agent(
    model="us.amazon.nova-lite-v1:0",
    tools=[file_read, calculator, python_repl],
    system_prompt="""You are a helpful assistant that can use tools to solve problems.
When using tools, always handle potential errors:
- For file operations, check if files exist before attempting to read them
- For mathematical operations, check for edge cases like division by zero
- When executing code, use try/except blocks to handle exceptions
"""
)

# Ask the agent to handle potential errors
response = error_handling_agent("""
Perform the following tasks, making sure to handle any potential errors:
1. Read the file 'nonexistent_file.txt' and summarize its contents
2. Calculate 5 divided by 0
3. Execute Python code to convert a string to an integer, but the string might not be a valid number
""")
```
